```python
from kafka import KafkaProducer
import json
import random
from datetime import datetime
import time

producer = KafkaProducer(
    bootstrap_servers='localhost:9092',
    value_serializer=lambda v: json.dumps(v).encode('utf-8')
)

print("Producer started")

for i in range(10):   # send only 10 records for notebook
    data = {
        "Patient_ID": random.randint(1, 100),
        "Timestamp": str(datetime.now()),
        "Sensor_ID": random.randint(1, 10),
        "Sensor_Type": "Heart Monitor",
        "Temperature_C": round(random.uniform(36, 39), 2),
        "Systolic_BP_mmHg": random.randint(100, 160),
        "Diastolic_BP_mmHg": random.randint(70, 100),
        "Heart_Rate_bpm": random.randint(60, 130),
        "Device_Battery_Level": random.randint(20, 100),
        "Target_Blood_Pressure": 120,
        "Target_Heart_Rate": 80,
        "Target_Health_Status": random.choice(["Stable", "Normal", "Critical"]),
        "Battery_Level": random.randint(20, 100)
    }

    producer.send('healthcare_stream', value=data)
    print("Sent:", data)

    time.sleep(2)
```
