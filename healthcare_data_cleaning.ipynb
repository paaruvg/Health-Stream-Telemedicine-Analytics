```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("healthcare_iot_target_dataset.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.columns = df.columns.str.strip()
```

```python
df.rename(columns={
    'Temperature (Â°C)': 'Temperature_C',
    'Systolic_BP (mmHg)': 'Systolic_BP',
    'Diastolic_BP (mmHg)': 'Diastolic_BP',
    'Heart_Rate (bpm)': 'Heart_Rate',
    'Device_Battery_Level (%)': 'Battery_Level'
}, inplace=True)
```

```python
df.columns
```

```python
df['Timestamp'] = pd.to_datetime(df['Timestamp'])
```

```python
df.info()
```

```python
df.dtypes
```

```python
numeric_cols = ['Temperature (°C)', 'Systolic_BP',
                'Diastolic_BP', 'Heart_Rate',
                'Battery_Level']

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
print(df.columns)
```

```python
df.isnull().sum()
```

```python

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].mean())
```

```python
df['Sensor_Type'] = df['Sensor_Type'].fillna('Unknown')
```

```python
df.isnull().sum()
```

```python
df = df[(df['Temperature (°C)'] >= 35) & (df['Temperature (°C)'] <= 42)]
```

```python
df = df[(df['Heart_Rate'] >= 40) & (df['Heart_Rate'] <= 180)]
```

```python
df = df[(df['Battery_Level'] >= 0) & (df['Battery_Level'] <= 100)]
```

```python
df.shape
```

```python
df = df.sort_values(by='Timestamp')
```

```python
df.reset_index(drop=True, inplace=True)
```

```python
df.to_csv("cleaned_healthcare_iot_dataset.csv", index=False)
```
